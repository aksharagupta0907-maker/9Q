```python

# Install the required libraries
!pip install transformers torch -q

from transformers import pipeline, set_seed

# Load the pre-trained GPT-2 model
generator = pipeline("text-generation", model="gpt2")

# Set random seed for reproducible results
set_seed(42)

# Reference poetry dataset
poetry_dataset = [
    "Roses are red, violets are blue,",
    "Sugar is sweet, and so are you.",
    "The moon glows bright in silent skies,",
    "A bird sings where the soft wind sighs."
]

# Prompt used for poem generation
prompt = (
    "Write a short, four-line poem inspired by these poems:\n"
    + "\n".join(poetry_dataset)
    + "\nNew poem:\n"
)

# Generate a new poem
result = generator(
    prompt,
    max_new_tokens=60,
    num_return_sequences=1,
    do_sample=True,
    temperature=0.8,
    pad_token_id=generator.tokenizer.eos_token_id
)

# Display the prompt and generated poem
print("Prompt Used:\n", prompt)
print("\nGenerated Poem:\n")
print(result[0]["generated_text"][len(prompt):])
```
